```python
import os
import pandas as pd
import cv2

RAW_DIR = r"C:\Users\patil\OneDrive\Desktop\lets win first hackathon\data\raw"
TRAIN_IMG_DIR = os.path.join(RAW_DIR, "train/train_images")
LABEL_OUT_DIR = r"C:\Users\patil\OneDrive\Desktop\lets win first hackathon\dataset\labels\train"

os.makedirs(LABEL_OUT_DIR, exist_ok=True)

# Class mapping
CLASS_MAP = {
    "bolt": 0,
    "locatingpin": 1,
    "nut": 2,
    "washer": 3
}
print(os.listdir(RAW_DIR))

df = pd.read_csv(os.path.join(RAW_DIR, "train_bboxes.csv"))

for image_name, group in df.groupby("image_name"):

    img_path = os.path.join(TRAIN_IMG_DIR, image_name)
    img = cv2.imread(img_path)

    if img is None:
        print(f"Image not found: {image_name}")
        continue

    h, w, _ = img.shape

    label_file = os.path.join(
        LABEL_OUT_DIR,
        image_name.replace(".png", ".txt")
    )

    with open(label_file, "w") as f:
        for _, row in group.iterrows():

            xmin, ymin, xmax, ymax = row[
                ["x_min", "y_min", "x_max", "y_max"]
            ]
            cls = CLASS_MAP[row["class"]]

            x_center = ((xmin + xmax) / 2) / w
            y_center = ((ymin + ymax) / 2) / h
            bw = (xmax - xmin) / w
            bh = (ymax - ymin) / h

            f.write(f"{cls} {x_center} {y_center} {bw} {bh}\n")

print("YOLO labels created successfully")
```
